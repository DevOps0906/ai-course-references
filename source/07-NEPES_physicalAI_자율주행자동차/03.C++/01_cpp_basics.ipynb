````python
from pathlib import Path
import os
import shlex
import subprocess

def find_course_dir():
    cwd = Path.cwd().resolve()
    candidates = [cwd, cwd / '03.C++']
    for parent in cwd.parents:
        candidates.extend([parent, parent / '03.C++'])
    for candidate in candidates:
        if (candidate / 'README.md').is_file() and (candidate / '01_cpp_basics').is_dir():
            return candidate
    raise FileNotFoundError('03.C++ 과정 폴더를 찾지 못했습니다.')

def run(args, cwd=None):
    command = [str(arg) for arg in args]
    print('$', shlex.join(command))
    result = subprocess.run(command, cwd=cwd or COURSE_DIR, text=True, capture_output=True, timeout=120)
    print(result.stdout, end='')
    if result.stderr: print(result.stderr, end='')
    result.check_returncode()
    return result

COURSE_DIR = find_course_dir()
LESSON_DIR = COURSE_DIR / '01_cpp_basics'
BUILD_DIR = COURSE_DIR.parent / 'Data' / 'build' / '03.C++' / LESSON_DIR.name
BUILD_DIR.mkdir(parents=True, exist_ok=True)

def compile_cpp(source, output_name, extra_flags=()):
    output = BUILD_DIR / output_name
    run(['g++', '-std=c++17', '-Wall', '-Wextra', '-Wpedantic', '-O0', '-g',
         *extra_flags, source, '-o', output])
    return output

os.chdir(COURSE_DIR.parent)
print('과정 폴더:', COURSE_DIR)
run(['g++', '--version'])
import sys
print('Python:', sys.executable)
````

컴파일 성공과 정답 통과는 다릅니다. 오류가 난 단계로 돌아가 수정·저장부터 다시 진행합니다.

C++는 보통 소스를 먼저 **컴파일** 해 기계가 실행할 파일을 만듭니다. 아래 한 줄에는 언어 버전, 경고, 입력 파일, 출력 파일이 모두 들어 있습니다.

```bash
g++ -std=c++17 -Wall -Wextra -Wpedantic main.cpp -o app
```

- `-std=c++17`: 이번 과정의 언어 규칙 선택
- `-Wall -Wextra -Wpedantic`: 실수를 조기에 찾는 경고
- `-o app`: 생성할 실행 파일 이름

컴파일 오류가 여러 개여도 먼저 **가장 위의 첫 오류** 부터 고치세요. 뒤의 오류는 첫 문제 때문에 연쇄적으로 생겼을 수 있습니다.

### 번역 단계와 오류의 위치

C++ 빌드는 한 번에 일어나는 것처럼 보여도 개념적으로 **전처리 → 컴파일 → 링크** 를 거칩니다. 전처리는 `#include`와 매크로를 펼쳐 번역 단위(translation unit)를 만들고, 컴파일은 각 번역 단위를 목적 파일로 바꾸며, 링크는 목적 파일과 라이브러리의 심볼을 연결해 실행 파일을 만듭니다. 따라서 문법·타입 문제는 주로 컴파일 오류로, 선언만 있고 정의가 없거나 필요한 라이브러리를 연결하지 않은 문제는 주로 링크 오류(`undefined reference`)로 나타납니다. 오류가 어느 단계에서 발생했는지 구분하면 확인할 범위를 빠르게 좁힐 수 있습니다.

### 가장 작은 프로그램 읽기

```cpp
#include <iostream>

int main() {
    std::cout << "안녕하세요!\n";
    return 0;
}
```

`#include`는 필요한 표준 라이브러리 선언을 가져오고, 운영체제는 `main`부터 실행합니다. 문장 끝의 세미콜론(`;`)과 중괄호 범위를 눈여겨보세요. `return 0`은 정상 종료를 뜻합니다.

### C++ 한 줄을 읽는 순서

처음에는 기호가 많아 보이지만 한 문장을 네 부분으로 나누면 됩니다.

```cpp
const double temperature{23.5};
```

1. `double`: 어떤 종류의 값을 저장하는가? 여기서는 소수가 있는 실수입니다.
2. `temperature`: 이 저장 공간을 이후 코드에서 부를 이름입니다.
3. `{23.5}`: 변수가 만들어질 때 넣는 최초 값입니다.
4. `const`: 초기화 뒤 이 이름을 통해 값을 바꾸지 않겠다는 약속입니다.

세미콜론은 문장의 끝이고, `//` 뒤는 사람을 위한 주석입니다. 컴파일러는 공백 모양보다 중괄호·괄호·세미콜론을 기준으로 구조를 읽습니다. 들여쓰기는 컴파일 규칙은 아니지만 사람이 중괄호 범위를 놓치지 않도록 도와줍니다.

Python과 달리 C++는 변수가 가질 타입을 컴파일할 때 결정합니다. 그래서 문자열을 정수 변수에 넣는 실수를 실행 전에 찾을 수 있습니다. 대신 정수끼리 나누면 소수부가 사라지는 것처럼 타입에 따른 연산 규칙을 의식해야 합니다.

### 예제 코드 해설 2 — 반복문과 누적 상태 (`02_loops.cpp`)

#### `for`가 합계를 만드는 과정

- 이 파일은 출력만 필요하므로 `<iostream>`만 포함합니다. `limit`은 고정된 반복 상한이고, `even_sum`은 반복할 때마다 바뀌는 **누적 변수** 라서 각각 `constexpr int`, `int`로 선언했습니다.
- `for (int number{1}; number <= limit; ++number)`는 `number`를 1로 만든 뒤, `number <= 10`인 동안 본문을 실행하고, 매 회 끝에 `number`를 1 증가시킵니다. 따라서 10도 포함됩니다.
- `number % 2`는 2로 나눈 나머지입니다. 나머지가 0인 2, 4, 6, 8, 10에서만 `even_sum += number`가 실행됩니다. 누적 값은 `0 → 2 → 6 → 12 → 20 → 30`으로 변하므로 첫 출력의 합은 30입니다.

#### `while`이 상태를 소진하는 과정

- `countdown`은 3에서 시작합니다. 조건 `countdown > 0`을 먼저 확인하고 현재 값을 출력한 다음 `--countdown`으로 하나 줄입니다. 상태는 `3 → 2 → 1 → 0`이 되고, 0에서는 조건이 거짓이라 반복이 끝난 뒤 `시작!`을 출력합니다.
- 출력 중 `' '`와 `'\n'`은 문자 하나라 작은따옴표를 사용합니다. `"시작!\n"`처럼 여러 글자는 문자열이므로 큰따옴표를 사용합니다.

> **흔한 오해:** `std::cout`은 반복 상태를 바꾸지 않습니다. `while`을 끝내는 핵심은 `--countdown`입니다. 이를 빠뜨리면 조건이 계속 참인 무한 반복이 됩니다. 또한 `%`는 퍼센트가 아니라 정수 나머지 연산자입니다.

#### 왜 미완성인데도 컴파일될까요?

`static_cast<void>(ticket_price);` 같은 문장은 값을 변환하거나 상태를 바꾸지 않습니다. 아직 TODO에서 사용하지 않은 변수를 '의도적으로 읽지 않았다'고 컴파일러에 알려 `-Wall -Wextra`의 미사용 경고를 피하는 scaffold입니다. 실제 계산에서 변수를 사용한 뒤 삭제해도 되고 남겨 두어도 동작에는 영향이 없습니다. `subtotal{0}`, `final_price{0}`도 같은 목적의 컴파일 가능한 자리표시자이지 정답이 아닙니다. 즉, **빌드 성공은 시작점** 이고 위 표의 값 흐름과 실행 결과까지 맞아야 완료입니다.

- **변경·부작용:** 명세에서 지목한 지역 변수와 소유 상태만 변경합니다. 실행은 stdout을 출력하고 컴파일러가 build artifact를 만들지만 별도 데이터 파일이나 GUI 상태는 변경하지 않습니다.
- **오류·경계:** 명시된 값·수명·소유권 경계를 지키고, 컴파일 오류·예외·assert/CTest 실패는 숨기지 않습니다. 실패한 build 또는 run은 0이 아닌 종료 코드로 드러나야 합니다.


### 노트북 편집 → 파일 저장 계약

````python
%%writefile 03.C++/01_cpp_basics/exercise/main.cpp
#include <iostream>

int main() {
    constexpr int ticket_price{12'000};
    constexpr int ticket_count{3};
    constexpr int membership_threshold{30'000};
    constexpr int discount_percent{10};
    const bool is_member{true};

    // TODO 1: 표 한 장의 가격과 장수를 사용해 결제 전 금액을 계산하세요.
    int subtotal{0};
    static_cast<void>(ticket_price);
    static_cast<void>(ticket_count);

    int discount{0};
    if (is_member && subtotal >= membership_threshold) {
        // TODO 2: 정수 연산으로 discount_percent만큼 할인액을 계산하세요.
        static_cast<void>(discount_percent);
    }

    // TODO 3: 결제 전 금액에서 할인액을 빼 최종 결제액을 계산하세요.
    const int final_price{0};

    std::cout << "결제 전: " << subtotal << "원\n";
    std::cout << "할인액: " << discount << "원\n";
    std::cout << "최종 결제액: " << final_price << "원\n";

    return 0;
}
````
