## 이 실습은 무엇을 위한 것인가요?

프로그램은 사람의 규칙을 입력에 따라 빠짐없이 적용하는 계산 절차입니다. 영화표나 배송비를 계산할 때 먼저 조건을 말로 설명하고, 그 조건을 변수·비교·분기·함수로 옮깁니다. 문법 이름을 외우는 것보다 어떤 입력이 어느 분기로 들어가 어떤 값을 반환하는지 추적하는 것이 목표입니다.

### 전체 과정에서 내 코드가 하는 일

```text
문제의 입력과 조건 읽기
  ↓
실행 전 결과 예측
  ↓
함수 TODO 구현
  ↓
정상·경계 입력 검증
  ↓
값과 판단 이유 기록
```

조건의 순서와 등호를 직접 작성합니다. 함수는 결과를 반환하고 호출한 쪽이 출력합니다. 한 예제의 값만 맞추는 상수 반환으로는 다른 입력의 규칙을 표현할 수 없습니다.

### 실행 전에 생각하고 결과로 설명하기

**질문:** 검증 한 번이 PASS이면 이 함수가 모든 현실 상황에 맞나요?

**설명:** 아닙니다. 제공한 계약과 사례를 통과한 것입니다. 경계값 양쪽과 잘못된 입력을 설명하고, 현실에서 추가될 조건은 별도 요구사항으로 정해야 합니다. print로 보이는 값과 함수가 반환한 값도 구분합니다.

이후의 TODO는 각 절에서 지정한 위치만 수정하고, 정답 해설의 이유를 자기 입력에 적용해 설명합니다. 예제 실행 성공·학생 함수 검증·실제 장치 관찰은 해당 절에서 요구하는 범위대로 나누어 기록합니다.

## 프로그래밍이 처음이라면 먼저 읽으세요

프로그램은 컴퓨터가 수행할 일을 정해진 문법으로 적은 **명령의 모음** 입니다. Python에서는 사람이 쓴 코드를 **Python 인터프리터** 가 위에서 아래로 읽어 실행합니다. 이 노트북의 **커널(kernel)** 은 그 인터프리터를 계속 실행해 두는 작은 작업 공간입니다. 한 셀에서 만든 변수와 함수가 다음 셀에서도 보이는 이유가 바로 커널이 상태를 기억하기 때문입니다.

처음에는 다음 용어만 구분하면 충분합니다.

| 용어 | 쉬운 뜻 | 예 |
|---|---|---|
| 값(value) | 프로그램이 다루는 실제 데이터 | `3`, `2.5`, `"Jetson"`, `True` |
| 변수(variable) | 값을 다시 찾기 위해 붙인 이름 | `temperature = 23.5` |
| 표현식(expression) | 계산되어 하나의 값이 되는 코드 | `temperature >= 30` |
| 문장(statement) | 대입·분기·반복처럼 수행할 일을 지시하는 코드 | `if alarm:` |
| 함수(function) | 입력을 받아 정해진 처리를 재사용하는 코드 묶음 | `len(values)` |
| 객체(object) | 값과 그 값이 할 수 있는 동작을 함께 보는 단위 | `values.append(3)`의 `values` |
| 오류/예외(error/exception) | 실행할 수 없는 이유를 알려 주는 진단 정보 | `NameError`, `TypeError` |

### 처음 보는 기호 읽기

| 기호 | 역할 | 읽는 예 |
|---|---|---|
| `# ...` | 그 줄의 설명인 주석. 실행되지 않음 | `# 온도 저장` |
| `=` | 오른쪽 값을 왼쪽 이름에 대입 | `x = 3` |
| `==` | 두 값이 같은지 비교 | `x == 3` |
| `()` | 함수 호출, 연산 순서 묶기, 튜플 표현 | `print(x)` |
| `[]` | 리스트 표현 또는 위치로 항목 선택 | `items[0]` |
| `{}` | 딕셔너리 표현 | `{"name": "A"}` |
| `:` | 조건·반복·함수·클래스 블록의 시작 | `if ready:` |
| 들여쓰기 | 어느 블록에 속한 문장인지 표시 | 보통 공백 4칸 |

코드의 `# TODO:`는 학습자가 채울 위치이고, `pass`는 아직 아무 동작도 하지 않는 빈 블록을 문법적으로 유지하는 문장입니다. `None`은 “아직 유효한 결과가 없음”을 나타내는 특별한 값으로 starter 코드에 사용됩니다.

### 오류 메시지를 읽는 순서

### 핵심 이론 · 이름 바인딩과 객체

Python의 변수는 값을 담는 상자라기보다 **객체를 가리키는 이름** 입니다. <code>b = a</code>는 객체를 복제하지 않고 같은 객체에 두 이름을 연결합니다. 이후 <code>a = 10</code>처럼 다시 대입하면 <code>a</code>의 연결만 바뀌지만, 리스트처럼 변경 가능한 객체를 <code>a.append(...)</code>로 수정하면 같은 객체를 가리키는 <code>b</code>에서도 변화가 보입니다. 반면 정수와 문자열은 불변 객체이므로 연산 결과가 필요할 때 새 객체가 만들어지고 이름이 다시 연결됩니다. **재대입** 과 **객체 내부 변경** 을 구분하면 함수 인자와 클래스 속성의 동작도 정확히 예측할 수 있습니다.

## 2. 조건문 (15–27분)

조건문은 질문의 답이 <code>True</code>인지 <code>False</code>인지에 따라 실행할 코드를 고릅니다.

- <code>if</code>: 첫 조건 검사
- <code>elif</code>: 앞 조건이 거짓일 때 추가 조건 검사
- <code>else</code>: 앞의 모든 조건이 거짓일 때 실행
- 비교: <code>==</code>, <code>!=</code>, <code>&gt;</code>, <code>&gt;=</code>, <code>&lt;</code>, <code>&lt;=</code>
- 논리: <code>and</code>, <code>or</code>, <code>not</code>

콜론(<code>:</code>)과 들여쓰기 4칸이 코드 블록의 경계를 만듭니다. 위에서 처음 참이 된 분기 하나만 실행됩니다.

````python
score = 83
was_absent = False

if was_absent:
    grade = '미응시'
elif score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
else:
    grade = '재학습'

print(f'{score}점의 결과: {grade}')
````

### 조건문을 손으로 추적하는 방법

조건은 위에서부터 확인합니다. 실행할 분기를 예측한 뒤 TODO와 경계 검사를 비교하세요.

현재 값은 `score == 83`, `was_absent is False`입니다.

1. `if was_absent`는 거짓이므로 `grade = '미응시'`를 건너뜁니다.
2. `score >= 90`도 거짓이므로 A 분기를 건너뜁니다.
3. `score >= 80`은 참이므로 `grade = 'B'`를 실행합니다.
4. 하나의 분기를 선택했으므로 뒤의 `elif`와 `else`는 더 검사하지 않습니다.

이처럼 조건문은 위에서 아래로 읽고 **처음 참이 된 위치에서 멈춘다** 고 생각하면 됩니다. `score`를 95, 75로 바꾸어 다시 실행하면 각각 어느 줄에서 멈추는지 설명해 보세요.

> **후속 과목 연결:** 물체 검출 신뢰도가 기준 이상일 때만 표시하거나, 센서 온도가 임계값 이상일 때 경보 명령을 보내는 코드가 같은 조건문 구조입니다.


````python
age = 20
has_ticket = True

print('성인인가?', age >= 19)
print('입장 가능한가?', age >= 19 and has_ticket)
print('표가 없는가?', not has_ticket)

# 범위 비교는 수학식처럼 이어 쓸 수도 있습니다.
temperature = 23
print('쾌적한 온도인가?', 18 <= temperature <= 26)
````

### `and`, `or`, `not`을 일상 문장으로 바꾸기

- `A and B`: A와 B가 **모두** 참이어야 참
- `A or B`: A와 B 중 **하나 이상** 참이면 참
- `not A`: A의 참/거짓을 뒤집음

현재 예제의 `age >= 19 and has_ticket`은 “성인이고 **동시에** 표도 있는가?”라는 질문입니다. 복합 조건이 헷갈리면 A와 B를 각각 `print`해서 확인한 뒤 결합하세요. `18 <= temperature <= 26`은 `18 <= temperature and temperature <= 26`을 Python답게 줄여 쓴 범위 검사입니다.

## 3. 반복문 (27–40분)

<code>for</code>는 정해진 값들을 하나씩 꺼내 반복할 때, <code>while</code>은 조건이 참인 동안 반복할 때 사용합니다.

- <code>range(start, stop, step)</code>에서 <code>stop</code>은 포함되지 않습니다.
- <code>enumerate(..., start=1)</code>는 순번과 값을 함께 줍니다.
- <code>break</code>는 반복을 즉시 끝내고, <code>continue</code>는 이번 회차만 건너뜁니다.
- 합계나 개수를 모으는 변수를 **누적 변수** 라고 합니다.

> **리스트 미리보기:** 여러 값을 순서대로 담는 <code>[]</code>를 리스트라고 합니다. <code>items.append(value)</code>는 끝에 값을 하나 추가합니다. 리스트 자체는 2교시에서 자세히 배웁니다.

````python
tasks = ['문법 읽기', '예제 실행', '실습 풀기']

for number, task in enumerate(tasks, start=1):
    print(f'{number}. {task}')

total = 0
for number in range(1, 6):  # 1, 2, 3, 4, 5
    total += number         # total = total + number

print('1부터 5까지의 합:', total)
````

첫 반복문에서 `enumerate(tasks, start=1)`은 매 회차에 순번과 항목을 한 쌍씩 줍니다.

| 회차 | `number` | `task` | 출력 |
|---:|---:|---|---|
| 1 | 1 | 문법 읽기 | `1. 문법 읽기` |
| 2 | 2 | 예제 실행 | `2. 예제 실행` |
| 3 | 3 | 실습 풀기 | `3. 실습 풀기` |

두 번째 반복문은 `total`을 반복 **전에** 0으로 준비하고, 각 숫자를 더한 결과를 다음 회차로 넘깁니다. 이런 변수를 누적 변수라고 합니다. 센서값 평균을 구하려면 합계와 개수를 누적하는 같은 방법을 씁니다.


````python
countdown = 3

while countdown > 0:
    print(countdown)
    countdown -= 1  # 조건이 언젠가 False가 되도록 값을 변경

print('시작!')
````

### `while` 반복의 핵심: 언젠가 끝나야 합니다

`while countdown > 0`은 매 회차 시작 때 조건을 다시 검사합니다. `countdown -= 1`이 3 → 2 → 1 → 0으로 값을 바꾸기 때문에 세 번 출력한 뒤 종료됩니다.

실수로 감소 문장을 지우면 조건이 계속 참이라 무한 반복합니다. 이때 JupyterLab 상단의 ■(Interrupt Kernel) 버튼을 누르고, 조건에 사용한 값이 반복 중에 어떻게 바뀌는지 확인하세요. 카메라나 ROS 노드는 계속 실행해야 하므로 `while`을 자주 쓰지만, 종료 조건과 자원 해제 코드를 함께 설계해야 합니다.

함수는 이름을 붙인 재사용 가능한 코드 블록입니다.

- **매개변수(parameter)** : 함수가 받을 입력의 이름
- **인자(argument)** : 호출할 때 실제로 전달하는 값
- **반환값(return value)** : 함수가 호출한 곳으로 돌려주는 결과
- **지역 변수(local variable)** : 함수 안에서 만든, 함수 밖에서는 직접 쓸 수 없는 변수

한 함수가 한 가지 일을 하게 만들고, 이름은 <code>동사_목적어</code>처럼 행동이 드러나게 지으면 좋습니다.

````python
def greet(name):
    """이름을 받아 인사 문자열을 반환한다."""
    return f'안녕하세요, {name}님!'


def rectangle_area(width, height):
    """직사각형의 넓이를 반환한다."""
    area = width * height  # area는 지역 변수
    return area


print(greet('서연'))
room_area = rectangle_area(4, 3)
print('방 넓이:', room_area)
````

## 선택/자습 확장 A — 클래스와 객체 (약 15–20분)

> **60분 수업 핵심을 완료했다면 여기서 먼저 핵심 정리를 확인해도 됩니다.** 이 절은 이후 코드를 더 깊이 이해하고 싶은 학습자를 위한 선택 자료입니다.

클래스는 서로 관련된 **상태(속성)** 와 **행동(메서드)** 을 묶는 설계도입니다. 클래스로 만든 실제 값을 객체 또는 인스턴스라고 합니다.

- `__init__`: 객체를 만들 때 처음 실행되는 초기화 메서드
- `self`: 지금 메서드를 실행 중인 객체 자신
- `self.balance`: 객체마다 따로 보관되는 인스턴스 속성
- `account.deposit(...)`: 객체의 메서드 호출

클래스 이름은 보통 `BankAccount`처럼 단어 첫 글자를 대문자로 씁니다. 예제와 `Book` TODO까지 진행하면 됩니다.


````python
class BankAccount:
    bank_name = '파이은행'  # 모든 인스턴스가 공유하는 클래스 속성

    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            return False
        self.balance += amount
        return True

    def withdraw(self, amount):
        if amount <= 0 or amount > self.balance:
            return False
        self.balance -= amount
        return True

    def summary(self):
        return f'{self.owner}: {self.balance:,}원'
````

### 클래스를 읽는 순서

클래스는 비슷한 데이터와 동작을 반복해서 만들기 위한 **설계도** 입니다. `BankAccount(...)`로 설계도에서 만든 실제 계좌 하나를 **인스턴스(instance)** 또는 객체라고 합니다.

1. `__init__`은 객체를 만들 때 한 번 실행되어 초기 상태를 저장합니다.
2. `self.owner`, `self.balance`는 각 객체가 가진 **인스턴스 속성** 입니다.
3. `deposit`, `withdraw`, `summary`는 객체가 수행할 **메서드** 입니다.
4. 첫 매개변수 `self`는 지금 그 메서드를 호출한 객체 자신입니다.
5. `bank_name`은 모든 계좌가 공유하는 **클래스 속성** 입니다.

`self.balance += amount`는 지역 변수만 바꾸는 것이 아니라 객체 안의 잔액 상태를 바꿉니다. 따라서 다음 셀에서 같은 객체의 `summary()`를 호출하면 바뀐 상태가 보입니다. 여러 로봇이나 카메라 상태를 각각 관리할 때도 인스턴스별 속성이 중요합니다.


````python
minji_account = BankAccount('민지', 50_000)
minji_account.deposit(20_000)
withdrawn = minji_account.withdraw(15_000)

print(BankAccount.bank_name)
print('출금 성공?', withdrawn)
print(minji_account.summary())
````

### 객체 실행 결과를 시간 순서로 읽기

처음 잔액 50,000원 → 20,000원 입금 후 70,000원 → 15,000원 출금 후 55,000원 순서입니다. `withdrawn`에는 잔액이 아니라 출금의 성공 여부 `True`가 들어갑니다. 하나의 메서드가 상태를 바꾸고 성공 여부를 반환할 수 있다는 점을 구분하세요.

메서드 호출 `minji_account.deposit(20_000)`은 개념적으로 `BankAccount.deposit(minji_account, 20_000)`처럼 현재 객체가 `self`로 전달된다고 이해할 수 있습니다. 실제 코드는 앞의 간결한 표기법을 사용합니다.

````python
class BookSolution:
    def __init__(self, title, total_pages):
        self.title = title  # 책마다 유지할 제목 상태
        self.total_pages = total_pages  # 진행률 계산의 기준이 되는 전체 쪽수
        self.current_page = 0  # 새 책은 읽지 않은 상태에서 시작

    def read(self, pages):
        if pages > 0:  # 0이나 음수 입력은 현재 상태를 바꾸지 않는다.
            self.current_page = min(self.current_page + pages, self.total_pages)  # 완독 지점에서 상한 적용

    def progress(self):
        return int(self.current_page / self.total_pages * 100)  # 소수 부분을 버린 정수 백분율


practice_book_solution = BookSolution('파이썬 첫걸음', 200)
practice_book_solution.read(35)
print(f'{practice_book_solution.title} 진행률: {practice_book_solution.progress()}%')
assert practice_book_solution.current_page == 35
assert practice_book_solution.progress() == 17
````

### 정답 해설 — 입력부터 결과까지

**객체 상태:** `__init__`는 책마다 제목·전체 쪽수·현재 읽은 쪽수를 준비합니다. `read`는 읽은 쪽수를 바꾸는 메서드이고 `progress`는 그 상태로 진행률을 계산하는 메서드입니다. `self`는 이번 호출의 책 객체를 가리킵니다.

10쪽 책에서 3쪽을 읽으면 30%, 그 뒤 20쪽을 더 읽어도 읽은 쪽수는 총 10쪽으로 제한되어 100%입니다. 음수 읽기를 허용하면 진행률이 거꾸로 가므로 검사합니다. 새 객체를 만들면 이전 책의 진행률을 이어받지 않습니다.

**확인 질문:** `progress()`를 여러 번 호출하는 것만으로 읽은 쪽수가 바뀌어야 할까요?

<summary><strong>정답 보기 — 통합 실습</strong></summary>

작은 판별은 함수에 맡기고, 범위를 기억하고 요약하는 책임은 클래스에 맡깁니다. 아래 접힌 코드를 한 줄씩 설명할 수 있는지 확인하세요.

````python
def is_even_solution(number):
    return number % 2 == 0  # 2로 나눈 나머지가 0이면 짝수


class NumberReportSolution:
    def __init__(self, start, end):
        self.start = start  # 요약할 포함 범위의 시작값 저장
        self.end = end  # 요약할 포함 범위의 끝값 저장

    def summarize(self):
        even_count = 0  # 조건을 만족한 수의 개수 누적
        even_sum = 0  # 조건을 만족한 수의 합계 누적

        for number in range(self.start, self.end + 1):  # end도 포함하도록 1을 더한다.
            if is_even_solution(number):  # 짝수 판별 책임은 보조 함수에 위임
                even_count += 1
                even_sum += number

        return f'짝수 {even_count}개, 합계 {even_sum}'  # 명세의 출력 형식으로 반환


report_solution = NumberReportSolution(1, 10)
summary_solution = report_solution.summarize()
print(summary_solution)
assert is_even_solution(8) is True
assert is_even_solution(7) is False
assert summary_solution == '짝수 5개, 합계 30'
````
