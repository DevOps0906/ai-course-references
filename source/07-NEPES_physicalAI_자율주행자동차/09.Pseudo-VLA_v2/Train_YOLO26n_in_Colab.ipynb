## 설명

### Ultralytics

Ultralytics는 AI 및 컴퓨터 비전(Computer Vision) 분야에서 널리 활용되는 최첨단 오픈소스 딥러닝 프레임워크와 모델을 개발하는 AI 기술 기업입니다. 특히 실시간 객체 탐지 알고리즘의 대표 주자인 **YOLO(You Only Look Once)** 시리즈(YOLOv5, YOLOv8 등)의 공식 메인테이너로 잘 알려져 있습니다.

**핵심 특징 및 장점**

* **쉬운 사용성:** 복잡한 PyTorch 코드를 몇 줄의 간결한 Python 코드 및 CLI 명령어로 추상화하여 모델 학습(Train), 검증(Val), 예측(Predict), 배포(Export)까지 한 번에 처리합니다.
* **경량화 및 고성능:** Nano부터 Extra Large까지 세분화된 모델 크기를 제공하여 스마트폰, 임베디드 기기(Edge AI), 서버 등 다양한 사양에서 실시간 추론을 지원합니다.
* **범용적 포맷 변환:** ONNX, TensorRT, CoreML, TFLite 등 다양한 타깃 배포 환경에 적합한 엔진 포맷으로 손쉽게 변환할 수 있습니다.

**지원하는 비전 태스크 비교**

| 태스크 | 주요 설명 | 출력 결과 |
| --- | --- | --- |
| **객체 탐지 (Detection)** | 이미지 내 사물의 위치와 종류를 동시에 식별 | 사각형 경계 상자 (Bounding Box) |
| **인스턴스 분할 (Segmentation)** | 객체의 정밀한 외곽 윤곽선을 픽셀 단위로 구분 | 마스크 (Mask) |
| **자세 추정 (Pose Estimation)** | 사람이나 객체의 주요 관절 위치 및 포즈 추적 | 핵심 포인트 (Keypoint) |
| **지향성 경계 상자 (OBB)** | 위성·항공 촬영 이미지 등 회전된 물체 정밀 측정 | 회전된 경계 상자 (Rotated Box) |
| **이미지 분류 (Classification)** | 이미지 전체의 카테고리 판별 | 라벨 및 확률값 |

컴퓨터 비전 서비스를 빠르게 제작하고 배포해야 하는 개발자 및 연구진에게 가장 높은 생산성을 제공하는 생태계입니다.

### YOLO26

**YOLO26**은 Ultralytics가 선보인 최신 실시간 컴퓨터 비전 모델로, 엔드투엔드 **NMS-Free(Non-Maximum Suppression Free)** 추론 구조를 도입하여 에지(Edge) 기기 및 실제 서비스 환경에서의 지연 시간을 대폭 줄인 프레임워크입니다. 후처리 연산의 병목을 없애고 깊이 추정(Depth Estimation) 등 지원 태스크를 확장한 점이 특징입니다.

**주요 특징 및 기술적 혁신**

* **NMS-Free 엔드투엔드 아키텍처:** 바운딩 박스를 정제할 때 필수적이었던 NMS 후처리 단계를 모델 구조 내부로 통합하여, CPU 및 임베디드 기기에서 발생하는 추론 병목을 크게 완화했습니다.
* **확장된 비전 태스크 지원:** 기존 객체 탐지, 인스턴스 세그멘테이션, 포즈 추정, OBB(지향성 경계 상자) 외에도 깊이 추정(Depth Estimation)과 시맨틱 세그멘테이션(Semantic Segmentation)이 신규 표준 라인업으로 포함되었습니다.
* **에지(Edge) AI 최적화:** 초경량 Nano(`yolo26n`)부터 고성능 Extra Large(`yolo26x`)까지 세분화된 스케일을 제공하며, ONNX, TensorRT, TFLite 등 온디바이스 배포 엔진으로의 가벼운 익스포트를 지원합니다.
* **통합 단일 API:** 기존 `ultralytics` 패키지 생태계와 100% 호환되므로 코드 변경을 최소화하면서 모델만 교체하여 사용할 수 있습니다.

**지원 태스크 및 모델 라인업**

| 태스크 | 주요 가중치 예시 | 주요 특징 및 출력 |
| --- | --- | --- |
| **객체 탐지 (Detection)** | `yolo26n.pt` ~ `yolo26x.pt` | 기본 NMS-Free 실시간 객체 탐지 |
| **인스턴스 분할 (Instance Seg.)** | `yolo26n-seg.pt` | 개별 사물의 픽셀 단위 경계 추출 |
| **시맨틱 분할 (Semantic Seg.)** | `yolo26n-sem.pt` | 픽셀 단위 범주 분류 (Cityscapes 학습 기반) |
| **깊이 추정 (Depth Estimation)** | `yolo26n-depth.pt` | 단일 이미지에서의 3D 깊이 맵 예측 |
| **자세 추정 (Pose Estimation)** | `yolo26n-pose.pt` | 사람 및 물체의 주요 관절(Keypoint) 추출 |
| **지향성 경계 상자 (OBB)** | `yolo26n-obb.pt` | 항공·위성 사진 등 회전된 물체 정밀 검출 |

복잡한 후처리 연산 알고리즘을 덜어내어 에지 디바이스나 실시간 로봇 제어 시스템 등 낮은 지연시간이 요구되는 실무 환경 구축에 가장 효과적인 옵션입니다.

### VOC dataset

PASCAL VOC(Visual Object Classes) 데이터셋은 컴퓨터 비전 알고리즘의 성능을 평가하기 위해 구축된 클래식 벤치마크 데이터셋입니다. 2005년부터 2012년까지 경진대회가 개최되었으며, 지금까지도 **VOC 2007**과 **VOC 2012**가 모델 연구 및 성능 비교의 표준 데이터셋으로 자주 활용됩니다.

**주요 20개 클래스 범주**

* **사람 (Person):** Person
* **동물 (Animal):** Bird, Cat, Cow, Dog, Horse, Sheep
* **탈것 (Vehicle):** Aeroplane, Bicycle, Boat, Bus, Car, Motorbike, Train
* **실내 물품 (Indoor):** Bottle, Chair, Dining table, Potted plant, Sofa, TV/monitor

**주요 지원 태스크**

| 태스크 | 주요 설명 | 어노테이션 형태 |
| --- | --- | --- |
| **객체 탐지 (Detection)** | 20개 클래스에 해당하는 객체의 위치와 종류 식별 | Bounding Box 좌표 (`xmin`, `ymin`, `xmax`, `ymax`) |
| **시맨틱 분할 (Segmentation)** | 픽셀 단위로 객체의 범주 및 외곽선 구분 | PNG 형태의 Color Mask |
| **이미지 분류 (Classification)** | 이미지 내 특정 객체의 존재 유무 판별 | Multi-label Class Flags |
| **행동 인식 (Action)** | 사람이 수행하는 행동(자전거 타기, 전화하기 등) 분류 | Action Label |

**데이터 포맷 및 주요 특징**

* **VOC XML 포맷:** 각 이미지의 어노테이션이 XML 파일 형태로 제공되며, 좌표 정보 외에도 객체가 부분적으로 가려졌는지(`truncated`), 판별하기 어려운지(`difficult`) 등의 메타데이터가 포함됩니다.
* **딥러닝 객체 탐지의 시초 벤치마크:** R-CNN, Fast R-CNN, SSD, 초기 YOLO(v1~v3) 등 주요 객체 탐지 모델들이 성능 검증을 위해 필수적으로 거쳐 간 기준 데이터셋입니다.
* **빠른 프로토타이핑에 적합:** MS COCO 데이터셋(80개 클래스, 33만 장 이상)에 비해 규모가 적당하여(VOC 2012 기준 약 11,530장 및 27,450개 객체), 모델 실험 및 빠른 가시화 학습용으로 널리 사용됩니다.

### VOC 데이터셋 분할

```python
import os
import shutil
from tqdm import tqdm
data_root = "/content/VOCData"
val_root = os.path.join(data_root, "val")
train_root = os.path.join(data_root, "train")
os.makedirs(val_root, exist_ok=True)
os.makedirs(train_root, exist_ok=True)

with open("/content/manifest.txt") as f:
  files = f.readlines()

for idx, img_path in tqdm(enumerate(files)):
  img_src = img_path.split('\n')[0]
  txt_src = os.path.splitext(img_src)[0] + ".txt"
  img_name = os.path.split(img_src)[-1]
  text_name = os.path.split(txt_src)[-1]
  if idx % 10 < 3:
    img_dst = os.path.join(val_root, img_name)
    text_dst = os.path.join(val_root, text_name)
  else:
    img_dst = os.path.join(train_root, img_name)
    text_dst = os.path.join(train_root, text_name)
  shutil.copy2(img_src, img_dst)
  shutil.copy2(txt_src, text_dst)
```

### 훈련을 위한 데이터셋 yaml 작성

```python
text_lines = """path: /content/VOCData # dataset root dir
train: train # train images (relative to 'path')
val: val # val images (relative to 'path')
test: #optional
names:
  0: aeroplane
  1: bicycle
  2: bird
  3: boat
  4: bottle
  5: bus
  6: car
  7: cat
  8: chair
  9: cow
  10: diningtable
  11: dog
  12: horse
  13: motorbike
  14: person
  15: pottedplant
  16: sheep
  17: sofa
  18: train
  19: tvmonitor"""
with open("/content/vocdata.yaml", 'w') as f:
  f.write(text_lines)
```

#### YOLO26n 모델 훈련

```python
from ultralytics import YOLO
# Load a model
model = YOLO("yolo26n.pt")
# Train the model
train_results = model.train(
  data="vocdata.yaml", # path to dataset YAML
  epochs=10, # number of training epochs
  imgsz=640, # training image size
  device=0, # device to run on, i.e. device=0 or device=0,1,2,3 or device=cpu
  batch=64,
  name="yolov26n_voc", # name of trained model
  save_dir="runs/detect/yolov26n_voc"
  )
```

#### OpenCV와 ONNX Runtime을 이용한 추론

```python
import cv2
import onnxruntime as ort
import numpy as np
# 1) 클래스 이름 로드
def get_class_names():
  class_names = [
    "aeroplane","bicycle","bird","boat","bottle","bus","car","cat","chair","cow",
    "diningtable","dog","horse","motorbike","person","pottedplant","sheep","sofa","train","tvmonitor"
  ]
  return class_names

# 2) ONNX Runtime 세션 생성
def get_session(model_path, providers=['CPUExecutionProvider', 'CUDAExecutionProvider']):
  session = ort.InferenceSession(model_path, providers=providers)
  input_name = session.get_inputs()[0].name # 보통 'images'
  return session, input_name

# 3) 이미지 로드 및 전처리 (640×640 리사이즈, RGB, 정규화)
def preprocessing(image_path):
  orig_image = cv2.imread(image_path)
  orig_h, orig_w = orig_image.shape[:2]
  img = cv2.resize(orig_image, (640, 640))
  img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB).astype(np.float32) / 255.0
  input_tensor = np.transpose(img_rgb, (2, 0, 1))[None, ...] # (1,3,640,640)
  return orig_image, orig_h, orig_w, input_tensor

# 4) 추론 실행
def inference_image(session, input_name, input_tensor):
  outputs = session.run(None, {input_name: input_tensor})
  return outputs[0][0] # outputs[0][0].shape == (num_detections, [x1, y1, x2, y2, confidence, class_id]) e.g. (1, 300, 6)

# 5) 결과 후처리
def get_detection_output(outputs, object_threshold):
  confidence = outputs[:, 4]
  indices = confidence > object_threshold
  if len(indices) == 0:
    return np.empty((0, 6)) # No detections after thresholding
  return outputs[indices, :]

# 6) 결과 시각화
def visualize(class_names, results, orig_image, orig_h, orig_w):
  result_image = orig_image.copy()
  for result in results:
    x1, y1, x2, y2 = result[0:4]
    x1, y1, x2, y2 = int(x1), int(y1), int(x2), int(y2)
    class_idx = int(result[5])
    score = "{:.2f}".format(result[4])
    cv2.rectangle(result_image,(int(x1),int(y1)),(int(x2),int(y2)),(255,0, 0),1)
    cv2.putText(result_image,class_names[class_idx]+' '+score,(x1,y1-17),
                cv2.FONT_HERSHEY_SIMPLEX,0.7,(255,0,0),1)
  return result_image
```

#### 추론

```python
model_path = 'runs/detect/yolov26n_voc/weights/best.onnx'
image_path = '/content/zidane.jpg'
object_threshold = 0.4
class_names = get_class_names()
session, input_name = get_session(model_path, ['CUDAExecutionProvider'])
orig_image, orig_h, orig_w, input_tensor = preprocessing(image_path)
outputs = inference_image(session, input_name, input_tensor)
detections = get_detection_output(outputs, object_threshold)
result_image = visualize(class_names, detections, orig_image, orig_h, orig_w)
cv2.imwrite('result.jpg', result_image)
```
