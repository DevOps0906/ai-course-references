### 1. 기초 복습: 환경 생성·reset·초기상태

- make_env는 과제와 환경 객체를 만듭니다.
- reset은 초기 상태를 적용하고 첫 관측을 반환합니다.
- initial_state는 준비된 배치 번호, seed는 난수 설정입니다.
- 동일 조건 재현과 서로 다른 배치 비교를 구분합니다.

환경 객체는 물리 모델·과제·관측 및 행동 인터페이스를 묶습니다. make_env는 libero_spatial suite에서 task_id와 초기상태 인덱스를 연결합니다. env.reset(seed=7)은 선택된 환경을 초기화하고 관측 obs와 부가 정보를 반환합니다. reset 자체가 정책 추론이나 과제 해결을 수행하지는 않습니다.

initial_state=0과 1은 준비된 초기상태 목록에서 어느 배치를 사용할지 정합니다. seed=7은 난수 경로를 고정하는 인자입니다. seed만 같다고 초기 배치 번호가 다른 두 장면이 같아지는 것은 아닙니다. 반대로 초기 배치를 바꾸려는 실험에서 seed까지 바꾸면 두 요인의 효과가 섞입니다.

같은 조건의 반복은 재현성을 살펴보는 활동이고 다른 초기 배치와의 비교는 입력 조건 변화에 대한 활동입니다. 두 카메라가 보는 방향이 다르므로 한 화면에서 작게 보이는 변화가 다른 화면에서는 더 뚜렷할 수 있습니다.


### 2. 실행 아키텍처: 노트북에서 결과 파일로

![실행 주체와 데이터 경계](../교재그림/guided-vla/detailed/day5-04.svg)

- run_lesson → run_or_read_lesson → 결과 경로 검사
- 완료 기록이 있으면 읽기; 없으면 run_episode 실행
- inspect → make_env/reset → NPZ·PNG·JSON
- FINISHED 창 닫기 → 노트북에서 두 영상 비교

모든 후속 노트북은 같은 SESSION 아래의 실행 이름으로 결과를 찾습니다. run_or_read_lesson은 result.json이 있으면 mode·seed·task·초기상태·horizon·스텝 한도를 확인한 뒤 읽습니다. 일치하는 완료 결과를 다시 여는 것은 새 reset이나 재측정이 아닙니다. 터미널의 재사용 안내를 먼저 읽습니다.

파일이 없을 때만 같은 Python으로 run_episode.py를 실행합니다. inspect는 초기 환경을 만든 뒤 initial-observation.npz, 두 카메라 PNG와 result.json을 저장하고 정책 로드 전에 끝납니다. GUI가 켜졌다면 마지막 장면을 확인하도록 대기하므로 창을 닫아야 subprocess가 반환하고 다음 셀이 진행됩니다.

새 실험이 필요하면 새 실행 이름을 사용합니다. 동일 폴더의 기존 기록을 지우거나 덮어쓰면 이전 비교 근거를 잃습니다. 재사용·새 실행·중단 재시도는 결과 이름과 상태로 구분합니다.


### 3. 코드 추적: 파일마다 다른 표현

- NPZ image/image2는 원본 RGB 배열입니다.
- PNG는 두 축 반전을 적용한 보기용 영상입니다.
- state는 make_batch가 만든 8차원 벡터입니다.
- result.json은 실행 조건·상태·시간을 기록합니다.

run_episode는 reset 직후 make_batch(obs, task)를 호출해 상태 벡터를 만들고 NPZ에 영상 원본과 state를 저장합니다. PNG 저장에서는 image[::-1,::-1].copy()로 방향을 바꿉니다. 따라서 NPZ의 raw와 PNG를 그대로 나란히 보고 상하·좌우가 다른 것을 모델 오류라고 생각하면 안 됩니다. 저장 목적과 변환 위치가 다릅니다.

NPZ는 이름 붙은 배열 묶음이므로 image, image2, state key를 통해 읽습니다. JSON은 배열 전체보다 실행 조건과 결과 상태를 읽기 좋게 저장합니다. 같은 실행 폴더의 파일들을 함께 사용해야 초기1 JSON에 초기0 영상을 붙이는 혼동을 막을 수 있습니다.

inspect 결과의 status=completed는 환경 관찰 작업이 끝났다는 뜻입니다. 이 모드는 행동을 적용하지 않으므로 작업 성공률에 포함하지 않습니다. mode와 termination_reason=inspection_only를 같이 확인합니다.


### 4. 초기0 반복·초기1 비교의 해석

- 과제·seed를 고정하고 초기 배치 번호만 바꿉니다.
- 두 영상과 픽셀 차이를 함께 봅니다.
- 픽셀 차이는 영상 차이이며 정책 성능 지표가 아닙니다.
- 같은 실행 재사용과 새 이름 재실행을 구분합니다.

초기0을 두 번 확인하는 활동은 동일 조건이 같은 관측을 만드는지 살펴보는 것입니다. 별도 실행 이름으로 실제 두 번 실행했는지, 이미 있는 파일을 재사용했는지에 따라 의미가 달라집니다. 셀 출력의 실행 경로 안내를 읽고 비교합니다. 조건을 같게 두어도 환경·렌더러에 따른 차이를 무조건 0으로 가정하지 않습니다.

초기1 비교에서는 물체·로봇의 배치가 어디서 달라졌는지 두 카메라로 확인합니다. 자동 픽셀 차이는 배열이 같거나 다름을 수치로 보여 주지만 변화의 원인이 물체 배치인지 시점인지 색인지 스스로 설명하지 않습니다. 이미지와 설정을 함께 읽어야 합니다.

직접 수정할 때는 make_env의 task_id, episode_index와 해상도, reset의 seed를 구분합니다. 해상도를 바꾸면 05의 입력 계약과 모델 전처리에도 영향을 주므로 단순히 GUI 크기를 바꾸는 일과 다릅니다.

| `FINISHED` | 종료 이유와 성공 여부 표시. 마지막 장면이며 더 움직이지 않음 |
| `INCOMPLETE / ERROR` | 미완료. 노트북의 원인 로그 확인 |

**정상 종료 화면을 관찰한 뒤 창을 닫으면 셀이 끝납니다.** 결과와 영상은 종료 화면 대기 전에 저장됩니다. 실행 도중 창을 닫으면 다음 안전한 처리 지점에서 사용자 중단으로 기록합니다. GPU 추론·모델 로딩 중에는 그 작업이 반환될 때까지 중단 반영이 늦을 수 있습니다. 창 조작은 표시용 복사본에만 적용되어 정책 카메라와 실제 실험 상태를 바꾸지 않습니다.

**기존 완료 결과를 읽으면 새 GUI는 열리지 않습니다.** 새 시연은 사용하지 않은 실행 이름으로 수행하고 변수·파일 경로도 함께 맞춥니다. GUI 사용 여부는 `result.json`의 `gui_enabled`에 남습니다. 화면 표시 비용이 있으므로 GUI 실행과 비GUI 실행의 시간을 같은 조건처럼 비교하지 않습니다.


**브라우저 탭을 닫아도 실험은 종료되지 않습니다.** 다시 같은 주소로 접속하면 진행 중 화면으로 돌아옵니다. FINISHED 후 화면 안의 MuJoCo 창을 닫아야 셀이 끝납니다. noVNC 연결과 탭은 다음 실험까지 유지하세요. 다른 학생 화면이 보이면 포트를 확인합니다.


## 선행 정보와 이 노트북의 입력·산출물

[03_container.ipynb의 「VLA 실습 준비 확인」](03_container.ipynb)에 전용 이미지·Data·커널 준비가 있으며 시작 절에 같은 명령을 반복 제공했습니다. [02_simulation_stack.ipynb의 도구 역할 설명](02_simulation_stack.ipynb)에서 LIBERO가 초기 배치와 성공조건을 정하고 MuJoCo가 두 가상 영상을 만드는 구조를 배웠습니다.

필요한 내용은 `reset`이 초기 관측을 만들고 `step(action)`이 행동 하나를 적용한다는 점입니다. 기본은 libero_spatial task0, seed7, initial_state=0입니다. 아래 설정 셀은 ROOT/SESSION/RUNS/run_lesson을 다시 정의하고 `inspect-01`이 없으면 정책 로드 없이 실제 reset을 수행합니다. 이전 노트북 산출물을 읽지 않습니다.

`RUNS/inspect-01/`에는 `result.json`(task·초기번호·seed·완료상태), `initial-observation.npz`(raw RGB 영상2와 상태8), `image.png`,`image2.png`(가로·세로 반전한 표시 영상)가 저장됩니다. 비교할 때 NPZ 원본과 PNG의 방향 차이를 시점 차이와 혼동하지 마세요.


## 개념

reset은 새 에피소드의 초기 관측을 얻는 동작이고 step은 행동 하나를 적용하는 동작입니다. 기준은 libero_spatial의 task 0입니다. 아래 실행은 정책을 로드하지 않고 실제 환경을 reset하여 카메라와 상태를 저장합니다.

seed는 난수 재현 설정이며 초기 배치 번호가 아닙니다. 초기 배치는 initial_state로 선택합니다. LeRobot 생성자의 episode_index를 사용하고 각 프로세스에서 한 번 reset합니다. 같은 번호·seed를 사용하면 같은 초기조건을 요청합니다. 동일 seed도 소프트웨어·장치 변화까지 완전한 재현을 보장하지 않습니다.

### 단계별 활동
