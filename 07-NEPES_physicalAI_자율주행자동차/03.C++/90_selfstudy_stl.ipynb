## 1. Python에서 무엇으로 옮길까?

| Python의 의도 | C++의 기본 선택 | 다른 선택이 필요한 때 |
|---|---|---|
| 크기가 변하는 순서 목록 `list` | `std::vector<T>` | 양 끝 삽입·삭제가 잦으면 `std::deque<T>` |
| 같은 타입, 고정 길이 | `std::array<T, N>` | 런타임 길이는 `vector` |
| 서로 다른 타입의 고정 묶음 `tuple` | `std::tuple<...>` | 필드 의미가 중요하면 이름 있는 `struct` |
| 키–값 `dict` | `std::unordered_map<K, V>` | 키 정렬·범위 탐색이 필요하면 `std::map<K, V>` |
| FIFO queue | `std::queue<T>` | 양 끝을 직접 다뤄야 하면 `std::deque<T>` |

> **중요:** Python `list`의 기본 대응은 C++의 `std::list`가 아니라 `std::vector`입니다. `std::list`는 노드가 흩어진 연결 리스트이며 임의 인덱싱을 제공하지 않습니다.

> **구분: 수업 후 복습·자습**

### 템플릿 표기와 컨테이너를 읽는 법

`std::vector<int>`에서 `vector`는 여러 타입에 재사용할 수 있는 컨테이너 설계이고, `<int>`는 이번 객체가 담을 원소 타입을 정합니다. 이를 템플릿 인자라고 합니다. `std::unordered_map<std::string, int>`에는 키 타입과 값 타입 두 개가 들어갑니다.

```text
std::unordered_map<std::string, int>
        컨테이너       키 타입      값 타입
```

C++ 컨테이너는 원소 타입도 컴파일할 때 정하므로 하나의 `vector<int>`에 문자열을 섞어 넣을 수 없습니다. 서로 다른 종류의 필드를 한 레코드로 묶으려면 `struct`나 `tuple`을 원소 타입으로 사용합니다.

반복자는 컨테이너 안의 위치를 나타내는 일반화된 도구입니다. `begin()`은 첫 원소 위치, `end()`는 마지막 원소 **다음** 위치입니다. `[begin, end)`처럼 끝을 포함하지 않는 범위 덕분에 빈 범위와 전체 길이를 일관되게 표현할 수 있습니다.

## 2. `std::vector`: 가장 먼저 고를 동적 배열

```cpp
std::vector<int> scores{80, 95, 72};
scores.push_back(88);

for (const int score : scores) {
    std::cout << score << '\n';
}
```

- `size()`, `empty()`: 크기와 비어 있는지 확인
- `push_back`, `emplace_back`: 뒤에 원소 추가
- `pop_back`: 마지막 원소 제거(값을 반환하지 않음)
- `operator[]`: 빠르지만 범위 검사 없음
- `at(index)`: 범위를 벗어나면 `std::out_of_range` 예외

원소를 읽기만 하면 `const auto&`, 수정하면 `auto&`, 작고 복사가 싼 숫자는 값 `auto`도 괜찮습니다.

### 반복자·참조 무효화(invalidation)

컨테이너의 원소 주소가 영원히 고정되는 것은 아닙니다. `vector`가 용량을 늘리며 재할당하면 기존 원소를 가리키던 **모든 반복자·포인터·참조** 가 무효가 되고, 재할당이 없더라도 삽입·삭제 지점 이후를 가리키던 것은 무효가 될 수 있습니다. 따라서 반복 중 `push_back`한 뒤 예전에 저장한 반복자를 계속 쓰지 말고, 변경이 끝난 뒤 `begin()`/`find()`로 다시 얻으세요. 반면 `map`의 삽입은 기존 원소의 반복자를 보통 유지하지만, 지운 원소를 가리키는 반복자는 어느 컨테이너에서든 다시 사용할 수 없습니다. 이 규칙은 값이 맞아 보여도 주소·별칭의 유효성이 이미 끝났을 수 있음을 뜻합니다.

## 5. `std::queue`: 먼저 온 값을 먼저 처리

```cpp
std::queue<std::string> jobs;
jobs.push("compile");
jobs.push("test");

while (!jobs.empty()) {
    const std::string current = jobs.front();
    jobs.pop();
    std::cout << current << '\n';
}
```

`front()`는 맨 앞 값을 읽고 `pop()`은 제거만 하며 값을 반환하지 않습니다. 따라서 읽은 뒤 제거하는 두 단계가 필요합니다. 빈 큐에서 `front()`나 `pop()`을 호출하면 안 되므로 반드시 `empty()`를 먼저 확인하세요. `queue`는 내부 컨테이너를 제한된 FIFO 인터페이스로 감싼 어댑터입니다. 단, `std::queue` 자체가 thread-safe한 것은 아닙니다. 이 예제는 한 스레드에서만 사용합니다. 여러 스레드가 공유한다면 `empty()` 검사·`front()` 읽기·`pop()` 제거를 같은 mutex 잠금 구간에서 수행하고 condition variable로 대기를 조율해야 합니다. 실제 스레드 생성과 대기 구현은 이 자습의 범위 밖입니다.

`const auto [order_id, item, quantity] = orders.front();`는 tuple의 세 원소를 **복사** 해 이름을 붙이는 C++17 구조적 바인딩입니다. 그다음 `pop()`으로 queue 원소를 제거해도 복사본은 살아 있습니다. `stock_it == stock.end()`이면 찾지 못했다는 뜻이고, 찾은 경우 `stock_it->second`가 해당 품목의 남은 수량입니다.
