````python
from pathlib import Path
import os
import shlex
import subprocess

def find_course_dir():
    cwd = Path.cwd().resolve()
    candidates = [cwd, cwd / '03.C++']
    for parent in cwd.parents:
        candidates.extend([parent, parent / '03.C++'])
    for candidate in candidates:
        if (candidate / 'README.md').is_file() and (candidate / '02_functions_structs_classes').is_dir():
            return candidate
    raise FileNotFoundError('03.C++ 과정 폴더를 찾지 못했습니다.')

def run(args, cwd=None):
    command = [str(arg) for arg in args]
    print('$', shlex.join(command))
    result = subprocess.run(command, cwd=cwd or COURSE_DIR, text=True, capture_output=True, timeout=120)
    print(result.stdout, end='')
    if result.stderr: print(result.stderr, end='')
    result.check_returncode()
    return result

COURSE_DIR = find_course_dir()
LESSON_DIR = COURSE_DIR / '02_functions_structs_classes'
BUILD_DIR = COURSE_DIR.parent / 'Data' / 'build' / '03.C++' / LESSON_DIR.name
BUILD_DIR.mkdir(parents=True, exist_ok=True)

def compile_cpp(source, output_name):
    output = BUILD_DIR / output_name
    run(['g++', '-std=c++17', '-Wall', '-Wextra', '-Wpedantic', '-O0', '-g',
         source, '-o', output])
    return output

print('교시 폴더:', LESSON_DIR)
run(['g++', '--version'])
os.chdir(COURSE_DIR.parent)

import sys
print('Python:', sys.executable)
````


### 선언 순서와 오버로딩

컴파일러는 호출 지점에서 함수의 이름과 타입을 이미 알아야 합니다. 정의가 아래쪽에 있다면 위에 **선언** 을 둡니다.

```cpp
double area(double radius);  // 선언
double area(double width, double height);  // 같은 이름, 다른 매개변수
```

함수 오버로딩은 매개변수의 **개수 또는 타입** 이 달라야 합니다. 반환 타입만 다르게 한 두 함수는 호출 시 구분할 수 없어 허용되지 않습니다. 기본 인자는 보통 선언에 한 번만 적습니다.

## 2. 값, 참조, `const` 참조 전달

| 형태 | 의미 | 알맞은 상황 |
|---|---|---|
| `void f(int value)` | 복사본 전달 | 작고 복사가 싼 값, 원본 변경 불필요 |
| `void f(int& value)` | 원본의 별칭 전달 | 호출자의 값을 바꿔야 함 |
| `void f(const std::string& text)` | 읽기 전용 별칭 | 큰 객체를 복사하지 않고 읽기 |

`const T&`는 '복사 비용은 피하지만 이 함수는 값을 바꾸지 않는다'는 계약입니다. 참조 매개변수 `T&`로 출력값을 여러 개 만들 수도 있지만, 처음에는 가능하면 명확한 반환값을 우선하세요.

### 45초 예측

```cpp
void add_one_copy(int value) { ++value; }
void add_one_ref(int& value) { ++value; }

int count{3};
add_one_copy(count);
add_one_ref(count);
```

마지막 `count`는 얼마일까요? 첫 호출은 복사본만 4가 되고, 두 번째 호출은 원본의 별칭을 바꿉니다. 함수 시그니처의 `&` 하나가 호출자의 상태 변경 여부를 결정합니다.

### 함수 호출과 전달 방식을 상자로 그리기

함수를 호출하면 매개변수 이름은 함수 안에서만 쓰는 지역 이름이 됩니다. 다음 그림처럼 생각할 수 있습니다.

```text
호출자 main의 count = 10

값 전달      update(count)      함수의 value = 10이라는 별도 복사본
참조 전달    update(count)      함수의 value ─────┐
                                              └── main의 count와 같은 객체
const 참조   inspect(count)     같은 객체를 복사 없이 보지만 이 경로로 변경 금지
```

작은 `int`는 값으로 전달해도 부담이 작습니다. 큰 `std::string`이나 `std::vector`를 읽기만 할 때는 `const T&`가 불필요한 복사를 피하면서 수정도 막아 줍니다. 호출자가 원본 변경을 기대할 때만 `T&`를 사용합니다. 참조 기호가 있다고 무조건 빠른 코드가 되는 것은 아니며, 함수의 의도를 드러내는 계약이라는 점이 더 중요합니다.

함수의 반환값과 출력도 구분하세요. `return total;`은 호출자에게 값을 돌려주고, `std::cout << total;`은 화면에 글자를 남깁니다. 출력만 한 함수의 결과를 다른 계산에 바로 사용할 수는 없습니다.

## 4. 클래스: 유효한 상태를 지키는 경계

C++에서 `struct`와 `class`는 거의 같은 기능을 가집니다. 기본 접근 권한만 `struct`는 `public`, `class`는 `private`입니다. 데이터 묶음은 `struct`, 규칙과 불변식(invariant)을 지키는 객체는 `class`로 시작하면 이해하기 쉽습니다.

```cpp
class BankAccount {
public:
    explicit BankAccount(int initial) : balance_{initial} {}
    bool withdraw(int amount);
    int balance() const { return balance_; }

private:
    int balance_;
};
```

생성자의 `: balance_{initial}`은 멤버 초기화 목록입니다. `balance() const`의 뒤쪽 `const`는 이 멤버 함수가 객체 상태를 바꾸지 않는다는 뜻입니다. 인자 하나짜리 생성자에는 의도하지 않은 암시적 변환을 막기 위해 `explicit`을 자주 사용합니다.

### 객체 수명과 멤버 초기화 순서

객체는 저장 공간이 확보되고 초기화가 완료된 뒤 사용할 수 있는 수명을 가집니다. 멤버는 초기화 목록에 적은 순서가 아니라 **클래스에 선언된 순서** 로 초기화되므로, 한 멤버의 초기값이 다른 멤버에 의존한다면 선언 순서까지 계약의 일부입니다. 자동 수명의 지역 객체는 블록을 벗어날 때 생성의 역순으로 소멸하며, 소멸 뒤 해당 객체를 가리키던 참조나 포인터를 사용하면 dangling 접근이 됩니다. 클래스가 생성자에서 불변식을 세우고 공개 멤버 함수가 이를 보존해야 하는 이유가 여기에 있습니다.

### 객체를 '상태 + 허용된 행동'으로 보기

구조체와 클래스는 여러 값을 하나의 새 타입으로 묶습니다. 이 수업에서는 클래스가 `private` 상태를 직접 노출하지 않고 `public` 멤버 함수로만 바꾸도록 하여 유효한 상태를 지키는 예를 다룹니다.

```text
Cart 객체
┌──────────────────────────────┐
│ private 상태                 │
│ item_count_, total_price_    │
│                              │
│ public 행동                  │
│ add(...), item_count(), ...  │
└──────────────────────────────┘
        ↑ 외부 코드는 public 경로로만 요청
```

객체가 지켜야 하는 규칙을 **불변식(invariant)** 이라고 합니다. 예를 들어 잘못된 수량이 들어왔을 때 개수만 늘고 금액은 그대로라면 두 상태가 서로 맞지 않게 됩니다. 멤버 함수는 입력을 먼저 검사하고, 유효할 때 관련 상태를 함께 변경해야 합니다.

멤버 함수 뒤의 `const`는 반환값이 상수라는 뜻이 아니라, 그 함수를 통해 객체 상태를 바꾸지 않겠다는 뜻입니다. 따라서 조회 함수와 변경 함수를 구분해서 읽을 수 있습니다.

```cpp
struct Product {
    std::string name;
    int unit_price;
};

int line_total(const Product& product, int quantity) {
    if (quantity <= 0 || product.unit_price <= 0) {
        return 0;
    }
    return product.unit_price * quantity;
}

public:

        const int amount{line_total(product, quantity)};

            return;
```
